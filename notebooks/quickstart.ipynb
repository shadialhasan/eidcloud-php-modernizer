Welcome to **eidcloud-php-modernizer** — an automated legacy PHP migrator and AST-driven modernizer upgrading PHP 5.x/7.x code to PHP 8.4+.

### Features demonstrated in this notebook:
1. Upgrading procedural `mysql_*` functions to modern **PDO** queries.
2. Migrating legacy PHP 4/5 class constructors to `__construct()`.
3. Upgrading properties to PHP 8 constructor property promotion.
4. Upgrading POSIX regex `ereg_*` to PCRE `preg_*`.
5. Upgrading `each()` loops to `foreach()`.
6. Upgrading verbose `isset()` ternaries to null coalescing `??`.
7. Upgrading `switch` statements to modern `match` expressions.
